```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv("../data/raw/Loan_Delinquent_Dataset (1).csv")

df.head()
```

```python
df["isDelinquent"].value_counts()
```

```python
df["isDelinquent"].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="isDelinquent")

plt.title("Loan Delinquency Distribution")
plt.xlabel("0 = Non-Delinquent | 1 = Delinquent")
plt.ylabel("Number of Customers")

plt.show()
```

```python
df["age"].value_counts()
```

```python



plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="age")

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="age", hue="isDelinquent")

plt.title("Age vs Loan Delinquency")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")

plt.legend(title="Delinquent")

plt.show()
```

```python
age_analysis = pd.crosstab(
    df["age"],
    df["isDelinquent"],
    normalize="index"
) * 100

age_analysis
```

```python
fico_analysis = pd.crosstab(
    df["FICO"],
    df["isDelinquent"],
    normalize="index"
) * 100

fico_analysis
```

```python
age_analysis.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Age-wise Delinquency Percentage")
plt.ylabel("Percentage")
plt.xlabel("Age Group")
plt.show()
```

```python
fico_analysis.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("FICO-wise Delinquency Percentage")
plt.ylabel("Percentage")
plt.xlabel("FICO Category")
plt.show()
```

```python
numeric_df = df.drop(columns=["ID"])
```

```python
encoded_df = pd.get_dummies(
    numeric_df,
    drop_first=True,
    dtype=int
)
```

```python
correlation = encoded_df.corr()

correlation
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```
